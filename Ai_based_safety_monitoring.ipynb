```python
from ultralytics import YOLO
import cv2

model = YOLO(
    r"C:\Users\RUDRANARAYAN DEBATA\runs\detect\train-16\weights\best.pt"
)

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame)

    if len(results[0].boxes) > 0:
        status = "SAFE"
    else:
        status = "UNSAFE"

    cv2.putText(
        frame,
        status,
        (20, 50),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 255, 0) if status == "SAFE" else (0, 0, 255),
        2,
    )

    annotated = results[0].plot()
    cv2.imshow("Safety Hat Detection", annotated)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
